## Evaluating an Embedding Model
Evaluating a model on MTEB follows a three step approach:

1) defining model
2) selecting the tasks
3) running the evaluation

Running MTEB on images requires you to install the optional dependencies using 

1. pip install mteb[image] 
2. uv add "mteb[image]

```python
import mteb
import pandas as pd
from sentence_transformers import SentenceTransformer
```

```python
# Specify the model that we want to evaluate

# model_name = "sentence-transformers/all-MiniLM-L6-v2"
model_name = "nvidia/Nemotron-3-Embed-8B-BF16"
# model = SentenceTransformer("sentence-transformers/LaBSE")
model = mteb.get_model(
    model_name
)  
```

```python
# specify what you want to evaluate it on
# if the model is not implemented in MTEB it will be eq. to SentenceTransformer(model_name)

# Select tasks
tasks = mteb.get_tasks(
        tasks=["Banking77Classification.v2"] # List of tasks to evaluate on
    )
```

While mteb.evaluate supports SentenceTransformers I recommend that the user use mteb.get_model to fetch the model as this prioritizes the implementation in mteb, which might not match 1-1 its SentenceTransformers implementation. For leaderboards results we see the mteb implementation as the reference implementation.

```python
# run the evaluation
model_results = mteb.evaluate(model, tasks=tasks)
print(model_results.model_name)
print(model_results.task_names)
```

```python
df = model_results.to_dataframe()
task_results = model_results.task_results

print(df) 
print(f"\nTask Results : {task_results}")
```
